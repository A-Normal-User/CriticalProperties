```python
import torch
import pandas as pd
import numpy as np
import GNN
from rdkit import Chem
model_Tc = GNN.GNN(node_dim = 24, edge_dim = 9, 
                conv_dim = 32, hidden_dim = 64, num_hidden_layers=8, dropout=0.0,
                num_gnn_layers = 8, gnn_heads = 4, pooling_heads = 2)
model_Tc.load_state_dict(torch.load('./model/GNN_Tc.pth', weights_only=True))
model_Tc.eval()

model_Pc = GNN.GNN(node_dim = 24, edge_dim = 9, 
                conv_dim = 32, hidden_dim = 64, num_hidden_layers=8, dropout=0.0,
                num_gnn_layers = 8, gnn_heads = 4, pooling_heads = 2)
model_Pc.load_state_dict(torch.load('./model/GNN_Pc.pth', weights_only=True))
model_Pc.eval()

def predict_Tc(smiles):
    GNNTc = GNN.GNN_Tc(model_Tc)
    GNNTc.eval()
    with torch.no_grad():
        pred = GNNTc(smiles)
    return pred

def predict_Pc(smiles):
    GNNPc = GNN.GNN_Tc(model_Pc)
    GNNPc.eval()
    with torch.no_grad():
        pred = GNNPc(smiles)
    return pred

df = pd.read_excel('./newexp/TEST_AB.xlsx')
need_drop = []
smiles = df['smiles'].to_list()
for i in range(len(df['smiles'])):
    try:
        GNN.mol_to_pyg(Chem.MolFromSmiles(smiles[i]))
        continue
    except Exception as e:
        print(f"Error processing SMILES {smiles[i]}: {e}")
        need_drop.append(i)
print(need_drop)
df = df.drop(need_drop)
smiles = df['smiles'].to_list()
TC = df['Tc (K)'].values.reshape(-1)
PC = df['Pc (bar)'].values.reshape(-1)
Tc_pred = predict_Tc(smiles).reshape(-1)
Pc_pred = predict_Pc(smiles).reshape(-1) / 100000
df['Tc_pred'] = Tc_pred
df['Pc_pred'] = Pc_pred 
df['Tc_error'] = np.abs(TC - Tc_pred) / TC * 100
df['Pc_error'] = np.abs(PC - Pc_pred) / PC * 100
df.to_excel('./newexp/GNN_pred.xlsx', index=False)
from sklearn.metrics import mean_squared_error, r2_score
tc_r2 = r2_score(TC, Tc_pred)
tc_rmse = np.sqrt(mean_squared_error(TC, Tc_pred))
tc_mae = np.mean(np.abs(TC - Tc_pred))
tc_aard = np.mean(np.abs(TC - Tc_pred) / TC) * 100
print(f'Tc R^2: {tc_r2:.4f}, RMSE: {tc_rmse:.4f}, MAE: {tc_mae:.4f}, AARD: {tc_aard:.4f}')
# 计算PC的R^2,RMSE,MAE,AARD
pc_r2 = r2_score(PC, Pc_pred)
pc_rmse = np.sqrt(mean_squared_error(PC, Pc_pred))
pc_mae = np.mean(np.abs(PC - Pc_pred))
pc_aard = np.mean(np.abs(PC - Pc_pred) / PC) * 100
print(f'Pc R^2: {pc_r2:.4f}, RMSE: {pc_rmse:.4f}, MAE: {pc_mae:.4f}, AARD: {pc_aard:.4f}')
```
