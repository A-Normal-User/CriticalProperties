```python
import pandas as pd
import numpy as np
import joblib

Tc_model = joblib.load('./model/GPR-Tc.pkl')
Pc_model = joblib.load('./model/GPR-Pc.pkl')
Pc_mean = pd.read_csv('./data/Pc_mean.csv', header=None).values.reshape(-1)
Tc_mean = pd.read_csv('./data/Tc_mean.csv', header=None).values.reshape(-1)
Tc_std = pd.read_csv('./data/Tc_std.csv', header=None).values.reshape(-1)
Pc_std = pd.read_csv('./data/Pc_std.csv', header=None).values.reshape(-1)

# 打开PC_feature3D_test.xlsx
df_Pc = pd.read_excel('./newexp/PC_feature3D_test.xlsx')
Pc_features = df_Pc.values[:, 4:]
Pc_features = (Pc_features - Pc_mean) / Pc_std
Pc_pred, std = Pc_model.predict(Pc_features, return_std=True)
# df_Pc只保留前四列
df_Pc = df_Pc.iloc[:, :4]
df_Pc['PC_pred'] = np.exp(Pc_pred) / 100000
df_Pc['Pc_low'] = np.exp(Pc_pred - 1.96 * std) / 100000
df_Pc['Pc_high'] = np.exp(Pc_pred + 1.96 * std) / 100000
df_Pc['AARD(%)'] = 100 * abs(df_Pc['PC_pred'] - df_Pc["Pc (bar)"]) / df_Pc["Pc (bar)"]
# 保存预测结果
df_Pc.to_excel('./newexp/GPR_3D_PC_TEST_AB.xlsx', index=False)

# 打开Tc_feature3D_test.xlsx
df_Tc = pd.read_excel('./newexp/Tc_feature3D_test.xlsx')
Tc_features = df_Tc.values[:, 4:]
Tc_features = (Tc_features - Tc_mean) / Tc_std
Tc_pred, std = Tc_model.predict(Tc_features, return_std=True)
# df_Tc只保留前四列
df_Tc = df_Tc.iloc[:, :4]
df_Tc['TC_pred'] = np.exp(Tc_pred)
df_Tc['Tc_low'] = np.exp(Tc_pred - 1.96 * std)
df_Tc['Tc_high'] = np.exp(Tc_pred + 1.96 * std)
df_Tc['AARD(%)'] = 100 * abs(df_Tc['TC_pred'] - df_Tc["Tc (K)"]) / df_Tc["Tc (K)"]
# 保存预测结果
df_Tc.to_excel('./newexp/GPR_3D_TC_TEST_AB.xlsx', index=False)
# 计算Tc和Pc预测的R^2,RMSE,MAE,AARD
from sklearn.metrics import mean_squared_error, r2_score
tc_r2 = r2_score(df_Tc["Tc (K)"], df_Tc['TC_pred'])
tc_rmse = np.sqrt(mean_squared_error(df_Tc["Tc (K)"], df_Tc['TC_pred']))
tc_mae = np.mean(np.abs(df_Tc["Tc (K)"] - df_Tc['TC_pred']))
tc_aard = np.mean(df_Tc['AARD(%)'])
print(f'Tc R^2: {tc_r2:.4f}, RMSE: {tc_rmse:.4f}, MAE: {tc_mae:.4f}, AARD: {tc_aard:.4f}')
pc_r2 = r2_score(df_Pc["Pc (bar)"], df_Pc['PC_pred'])
pc_rmse = np.sqrt(mean_squared_error(df_Pc["Pc (bar)"], df_Pc['PC_pred']))
pc_mae = np.mean(np.abs(df_Pc["Pc (bar)"] - df_Pc['PC_pred']))
pc_aard = np.mean(df_Pc['AARD(%)'])
print(f'Pc R^2: {pc_r2:.4f}, RMSE: {pc_rmse:.4f}, MAE: {pc_mae:.4f}, AARD: {pc_aard:.4f}')
```
