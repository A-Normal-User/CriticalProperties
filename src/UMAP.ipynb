# PC

## training and validation set

```python
import pandas as pd
import numpy as np
import umap
import matplotlib.pyplot as plt
df = pd.read_csv('../database/Crit-Pc.csv')
df_meanstd = pd.read_csv('../database/Crit-Pc-MeanStd.csv')
mean = df_meanstd['mean'].values[:76-11]
std = df_meanstd['std'].values[:76-11]
Fea = (df.iloc[:, 2:].values - mean) / std
training = df['Validation'] == 0
validation = df['Validation'] == 1

X_scaled = Fea
X_train = X_scaled[training]
X_val = X_scaled[validation]
reducer = umap.UMAP(
    n_components=2,
    a=1.0,
    b=0.8,
    n_neighbors=15,
    metric='euclidean',
    init='random',
    random_state=42,
    n_jobs=1
)
embedding = reducer.fit_transform(X_train)
plt.figure(figsize=(7, 5))
scatter = plt.scatter(
    embedding[:, 0], embedding[:, 1],
    color='gray', s=16, alpha=0.6,
    label='Training'
)
embedding2 = reducer.transform(X_val)
scatter = plt.scatter(
    embedding2[:, 0], embedding2[:, 1],
    zorder=3,
    color='blue',
    s=24, alpha=0.6,
    marker='x',
    label='Validation'
)
plt.grid(True, linestyle='--', alpha=0.3)
plt.legend(fontsize=18, loc='lower right')
plt.clim(0, None)
plt.tight_layout()
plt.show()
```

## training and test set
