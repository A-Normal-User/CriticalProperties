# Build Feature

## PC

```python
import numpy as np
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors
from skfp.fingerprints import PubChemFingerprint
from mordred import Calculator, descriptors
def make_PC_feature(smiles, names):
    fea_rdkit = ['MinEStateIndex', 'qed', 'SPS', 'BalabanJ', 'Chi0n', 'Kappa1', 'Kappa3', 'VSA_EState3', 'VSA_EState8', 'NHOHCount', 'NumHDonors', 'Phi', 'MolLogP']
    rdkit2D = {}
    pubchem_fp = PubChemFingerprint(count=True)
    for i in range(len(smiles)):
        name = names[i]
        smile = smiles[i]
        mol = Chem.MolFromSmiles(smile)
        rdkit2D[name] = {}
        rdkit2D[name]['trcid'] = name
        rdkit2D[name]['smiles'] = smile
        for Descriptor, func in Descriptors.descList:
            if Descriptor in fea_rdkit:
                try:
                    rdkit2D[name][Descriptor] = func(mol)
                except:
                    pass
        fingerprint = pubchem_fp.transform([smile])[0]
        rdkit2D[name]['PubChemFP_fp_0'] = fingerprint[0]
        rdkit2D[name]['PubChemFP_fp_9'] = fingerprint[9]
    calc_all = Calculator(descriptors, ignore_3D=False)
    fea = ['nAtom', 'ATS0se', 'ATS2se', 'ATS6se', 'ATS0are', 'ATS5p', 'ATS6p', 'ATS0i', 'ATS2i', 'ATS6i', 'AATS0v', 'ATSC1dv', 'ATSC2dv', 'ATSC0v', 'ATSC1v', 'ATSC1pe', 'ATSC1are', 'ATSC1p', 'ATSC0i', 'AATSC0i', 'BCUTdv-1l', 'BCUTse-1l', 'BCUTare-1l', 'BalabanJ', 'SpMAD_DzZ', 'SpAbs_Dzse', 'SpMAD_Dzse', 'SpMAD_Dzpe', 'SpMAD_Dzare', 'SpAbs_Dzi', 'SpDiam_Dzi', 'SpMAD_Dzi', 'Sse', 'Spe', 'Sare', 'Sp', 'Si', 'SsCH3', 'ETA_dEpsilon_D', 'fMF', 'CIC0', 'ZMIC1', 'FilterItLogS', 'VMcGowan', 'AMID_h', 'MID_C', 'AMID_O', 'apol', 'bpol', 'TopoPSA']
    selected_desc = [d for d in calc_all.descriptors if str(d) in fea]
    mols = [Chem.MolFromSmiles(smile) for smile in smiles]
    calc = Calculator(selected_desc, ignore_3D=False)
    mordred = calc.pandas(mols, quiet=True)
    df_ = pd.DataFrame(rdkit2D).T
    df_.reset_index(drop=True, inplace=True)
    df_.columns = ['Name', 'smiles', 'RDKit2D_MinEStateIndex', 'RDKit2D_qed', 'RDKit2D_SPS', 'RDKit2D_BalabanJ', 'RDKit2D_Chi0n', 'RDKit2D_Kappa1', 'RDKit2D_Kappa3', 'RDKit2D_VSA_EState3', 'RDKit2D_VSA_EState8', 'RDKit2D_NHOHCount', 'RDKit2D_NumHDonors', 'RDKit2D_Phi', 'RDKit2D_MolLogP', 'PubChemFP_fp_0', 'PubChemFP_fp_9']
    mordred.columns = ['Mordred2D_' + col for col in mordred.columns]
    return pd.concat([df_, mordred], axis=1)
names = []
smiles = []
for i in range(1, 201):
    names.append(f'C{i}')
    smiles.append('C' * i)
feature_df = make_PC_feature(smiles, names)
feature_df.to_csv('./homo/PC_feature.csv', index=False)
```

```python
# TC
import GNN
import numpy as np
import pandas as pd
import torch
from rdkit import Chem
torch.set_num_threads(16)
seed_list=[42, 403138084, 573583737, 604087939, 643026125, 661892772, 866264613, 1509598813, 1605674636, 1687343257]
df = pd.read_csv('./homo/TC_feature.csv')
need_drop = []
for index, row in df.iterrows():
    try:
        GNN.mol_to_pyg(Chem.MolFromSmiles(row['smiles']))
    except:
        need_drop.append(index)
df = df.drop(need_drop).reset_index(drop=True)
X_train = df['smiles'].to_list()
T_mean = 6.454735946249426 
T_std = 0.22434263421735687
input_loader = GNN.preprocess(X_train)

models = []
hidden_dim = 80
num_gnn_layers = 2
num_hidden_layers = 7
gnn_heads = 5
pooling_heads = 2
dropout = 0.044634731325588844
for seed in seed_list:
    model = GNN.GNN(node_dim = 38, edge_dim = 9, 
                conv_dim = 32, hidden_dim = hidden_dim, num_hidden_layers=num_hidden_layers, dropout=dropout,
                num_gnn_layers = num_gnn_layers, gnn_heads = gnn_heads, pooling_heads = pooling_heads)
    model.load_state_dict(torch.load(f'../model/GNN_TC_{seed}.pth', weights_only=True, map_location=torch.device('cpu')))
    model.eval()
    models.append(model)

results = []
with torch.no_grad():
    for model in models:
        y_pred = []
        for X in input_loader:
            y_pred.extend(model(X.x, X.edge_index, X.edge_attr, X.numHDonors, X.numHAcceptors, X.batch).numpy())
        results.append(np.array(y_pred))
results = np.array(results)
results = results * T_std + T_mean
y_pred_mean = np.mean(results, axis=0).flatten()
y_std = np.std(results, axis=0)
df['TC_pred'] = np.exp(y_pred_mean)
df['TC_std'] = y_std
df = df[['Name', 'TC_pred', 'TC_std']]
df.to_csv('./homo/GNN-TC.csv', index=False)
```

```python
import numpy as np
import pandas as pd
import torch
from model import Crit_model
torch.set_num_threads(16)
seed_list=[42, 403138084, 573583737, 643026125, 866264613, 1509598813, 1605674636, 1687343257]
df = pd.read_csv('./homo/TC_feature.csv')
df_meanstd = pd.read_csv('../database/Crit-Tc-MeanStd.csv')
mean = df_meanstd['mean'].values[:92-11]
std = df_meanstd['std'].values[:92-11]
X_train = df.iloc[:, 2:].astype(np.float32)
X_train = (X_train - mean) / std
X_train = X_train.drop(columns=['Mordred2D_SpAbs_Dzp']).values
T_mean = 6.454735946249426 
T_std = 0.22434263421735687

models = []
hidden_dim = 56
hidden_num = 4
hidden_type = "2_2"
activation = "sigmoid"
dropout = 0.0
for seed in seed_list:
    model = Crit_model(input_size=X_train.shape[1], 
                       output_size=1,
                       hidden_size=hidden_dim, 
                       hidden_num=hidden_num, 
                       hidden_type=hidden_type, 
                       activation=activation, 
                       dropout=dropout)
    model.load_state_dict(torch.load(f'../model/ANN_dsTC2D_{seed}.pth', weights_only=True, map_location=torch.device('cpu')))
    model.eval()
    model = torch.jit.script(model)
    models.append(model)

X_train = torch.from_numpy(X_train).float()
results = []
with torch.no_grad():
    for model in models:
        y_pred = model(X_train).numpy()
        results.append(y_pred)
results = np.array(results)
results = results * T_std + T_mean
y_pred_mean = np.mean(results, axis=0).flatten()
y_std = np.std(results, axis=0)
df['TC2D_pred'] = np.exp(y_pred_mean)
df['TC2D_std'] = y_std
df = df[['Name', 'TC2D_pred', 'TC2D_std']]
df.to_csv('./homo/dsANN-TC.csv', index=False)
```
